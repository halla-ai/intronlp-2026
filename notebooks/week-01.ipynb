### 1-2. 확률표에서 다음 단어 뽑기

강의 2에서 본 두 확률표를 그대로 옮겼습니다. **숫자는 설명하려고 정한 예시입니다.**

- `혼저 ___` 의 표는 한 후보에 몰려 있습니다 (뾰족한 표)
- `제주에서 먹고 싶은 것은 ___` 의 표는 넓게 퍼져 있습니다 (평평한 표)

각 표에서 다음 단어를 10번씩 뽑아 몇 번씩 나왔는지 셉니다. 셀을 다시 실행할 때마다 결과가 조금씩 바뀝니다.

```python
import random
from collections import Counter

sharp = {"옵서예": 0.90, "오세요": 0.05, "그 밖의 말": 0.05}
flat = {"고기국수": 0.25, "흑돼지": 0.24, "갈치조림": 0.18, "전복죽": 0.12, "그 밖의 말": 0.21}

def draw(table, times=10):
    picks = random.choices(list(table), weights=list(table.values()), k=times)
    return Counter(picks)

print("혼저 ___                  ", draw(sharp))
print("제주에서 먹고 싶은 것은 ___", draw(flat))
```
